---## Bölüm 1 — Series**Soru 1.1:** Bir kırtasiyedeki ürünlerin stok adetlerini tutan bir Series oluşturun: index = `['Kalem','Defter','Silgi','Cetvel']`, değerler = `[300, 150, 80, 45]`.**Soru 1.2:** `'Silgi'` ürününün stok adedini index adıyla seçip yazdırın.**Soru 1.3:** Stoğu en az olan ürünü `.idxmin()` ile bulun.**Soru 1.4:** Tüm ürünlerin toplam stok adedini `.sum()` ile hesaplayın.

```python
# 1.1: Series Oluşturma
stok = pd.Series([300, 150, 80, 45], index=['Kalem', 'Defter', 'Silgi', 'Cetvel'])
print(stok)

# 1.2: Index Adıyla Eleman Seçme
print("Silgi Stoğu:", stok['Silgi'])

# 1.3: En Az Stoğu Olan Ürünü Bulma (.idxmin)
en_az_urun = stok.idxmin()
print("En az stoklu ürün:", en_az_urun)

# 1.4: Toplam Stok Adedi (.sum)
toplam_stok = stok.sum()
print("Toplam Stok:", toplam_stok)
```

---## Bölüm 4 — Veriyi İncelemeAşağıdaki hücreyi çalıştırarak bir e-ticaret sipariş verisi oluşturun:

```python
df = pd.DataFrame({
    "siparis_no": [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "kargo_suresi": [2, 5, 3, np.nan, 1, 4, 6],
    "urun_adedi": [1, 3, 2, 5, 1, 2, 4],
    "toplam_tutar": [150.0, 480.5, 220.0, 610.0, 90.0, 310.0, 720.0]
})
```

**Soru 4.1:** `.head(3)` ve `.tail(2)` ile ilk 3 ve son 2 siparişi görüntüleyin.**Soru 4.2:** `.info()` ile hangi sütunda eksik veri olduğunu tespit edin.**Soru 4.3:** `.describe()` ile kargo süresi, ürün adedi ve toplam tutarın istatistiklerini inceleyin.

```python
# 4.1: İlk 3 ve Son 2 Satır
print("--- İlk 3 ---")
print(df.head(3))
print("\n--- Son 2 ---")
print(df.tail(2))

# 4.2: Veri Özeti (.info)
df.info()  # kargo_suresi sütununda 6 non-null değer olduğu için 1 eksik (NaN) veri var.

# 4.3: İstatistiksel İnceleme (.describe)
print(df.describe())
```

---## Bölüm 5 — Sütun ve Satır SeçmeYukarıdaki sipariş `df`'ini kullanarak:**Soru 5.1:** Sadece `siparis_no` ve `toplam_tutar` sütunlarını seçip yazdırın.**Soru 5.2:** `.loc[]` ile 0'dan 3'e kadar olan (0-3 dahil) satırları seçin.**Soru 5.3:** `.iloc[]` ile ilk 4 satırın son 2 sütununu seçin.

```python
# 5.1: Belli Sütunları Seçme
print(df[['siparis_no', 'toplam_tutar']])

# 5.2: .loc ile Etiket/Index Bazlı Seçim (0 ve 3 dahil)
print(df.loc[0:3])

# 5.3: .iloc ile Indeks Bazlı Seçim (ilk 4 satır, son 2 sütun)
print(df.iloc[0:4, -2:])
```

---## Bölüm 6 — Filtreleme ve Yeni Sütun**Soru 6.1:** Toplam tutarı 300 TL'nin üzerinde olan siparişleri filtreleyin.**Soru 6.2:** Ürün adedi 2'den fazla VE toplam tutar 200 TL'den fazla olan siparişleri bulun (`&` kullanın).**Soru 6.3:** `birim_fiyat` adında yeni bir sütun ekleyin; değeri `toplam_tutar / urun_adedi` olsun.**Soru 6.4:** `apply` ve `lambda` kullanarak kargo süresi 3 günden fazlaysa "Yavaş", değilse "Hızlı" yazan bir `kargo_durumu` sütunu oluşturun.

```python
# 6.1: Tutara göre filtreleme
print(df[df['toplam_tutar'] > 300])

# 6.2: Çoklu şart filtreleme (&)
print(df[(df['urun_adedi'] > 2) & (df['toplam_tutar'] > 200)])

# 6.3: Matematiksel Operasyonla Yeni Sütun
df['birim_fiyat'] = df['toplam_tutar'] / df['urun_adedi']

# 6.4: apply ve lambda Kullanımı
df['kargo_durumu'] = df['kargo_suresi'].apply(lambda x: "Yavaş" if x > 3 else "Hızlı")
print(df)
```

---## Bölüm 7 — Sıralama**Soru 7.1:** Siparişleri `toplam_tutar` sütununa göre büyükten küçüğe sıralayın.**Soru 7.2:** Siparişleri önce `urun_adedi`, sonra `kargo_suresi` sütununa göre (ikisi de küçükten büyüğe) sıralayın (bir listede iki sütun verin).

```python
# 7.1: Tek Sütuna Göre Azalan Sıralama
df_sirali1 = df.sort_values(by='toplam_tutar', ascending=False)
print(df_sirali1)

# 7.2: Çoklu Sütuna Göre Artan Sıralama
df_sirali2 = df.sort_values(by=['urun_adedi', 'kargo_suresi'], ascending=[True, True])
print(df_sirali2)
```

---## Bölüm 8 — Eksik Veri (NaN)**Soru 8.1:** `kargo_suresi` sütununda kaç eksik değer olduğunu bulun.**Soru 8.2:** Eksik kargo süresini sütunun **medyanı** ile doldurup `df_dolu` adlı yeni bir DataFrame oluşturun.**Soru 8.3:** Eksik değer içeren satırı silerek `df_temiz` adlı yeni bir DataFrame oluşturun ve satır sayısının değiştiğini `.shape` ile doğrulayın.

```python
# 8.1: Eksik Değer Sayısı
eksik_sayisi = df['kargo_suresi'].isna().sum()
print("Eksik kargo süresi sayısı:", eksik_sayisi)

# 8.2: Medyan ile Doldurma (.fillna)
df_dolu = df.copy()
medyan_deger = df_dolu['kargo_suresi'].median()
df_dolu['kargo_suresi'] = df_dolu['kargo_suresi'].fillna(medyan_deger)

# 8.3: Eksik Satırı Silme (.dropna)
df_temiz = df.dropna()
print("Orijinal Shape:", df.shape)
print("Temizlenmiş Shape:", df_temiz.shape)
```

---## Bölüm 9 — Merge (Tabloları Birleştirme)Aşağıdaki iki tabloyu kullanın:

```python
musteriler = pd.DataFrame({
    "musteri_id": [1, 2, 3, 4],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya"]
})
siparisler = pd.DataFrame({
    "musteri_id": [1, 2, 5],
    "urun": ["Kulaklık", "Şarj Kablosu", "Powerbank"],
    "tutar": [450, 90, 320]
})
```

**Soru 9.1:** `musteriler` ve `siparisler` tablolarını **inner join** ile birleştirin — kaç müşterinin siparişi eşleşiyor?**Soru 9.2:** Aynı tabloları **left join** ile birleştirin — hiç siparişi olmayan müşteri(ler) sonuçta ne görünüyor?**Soru 9.3:** Aynı tabloları **outer join** ile birleştirip, hem eşleşmeyen müşterileri hem eşleşmeyen siparişleri gözlemleyin.

```python
# 9.1: Inner Join (Sadece Kesişim)
df_inner = pd.merge(musteriler, siparisler, on='musteri_id', how='inner')
print("--- Inner Join ---")
print(df_inner)  # 2 müşterinin (1 ve 2) siparişi eşleşir.

# 9.2: Left Join (Sol Tablo Tamam, Sağ Eşleşen)
df_left = pd.merge(musteriler, siparisler, on='musteri_id', how='left')
print("\n--- Left Join ---")
print(df_left)  # Siparişi olmayan 3 ve 4 id'li müşterilerin ürün/tutar kısımları NaN görünür.

# 9.3: Outer Join (Tüm Kayıtlar)
df_outer = pd.merge(musteriler, siparisler, on='musteri_id', how='outer')
print("\n--- Outer Join ---")
print(df_outer)  # Eşleşmeyen 5 id'li sipariş ve eşleşmeyen 3-4 id'li müşterilerin hepsi gelir.
```

---## Bölüm 10 — Concat**Soru 10.1:** `hafta1 = pd.DataFrame({"gun":["Pzt","Sal"], "adet":[20,35]})` ve `hafta2 = pd.DataFrame({"gun":["Çar","Per"], "adet":[28,40]})` tablolarını `ignore_index=True` ile alt alta birleştirin.**Soru 10.2:** `gelir = pd.DataFrame({"toplam_gelir":[1000,1750]})` tablosunu, `hafta1` tablosuyla `axis=1` kullanarak yan yana birleştirin.

```python
# 10.1: Dikey Birleştirme (Satır Bazlı)
hafta1 = pd.DataFrame({"gun":["Pzt","Sal"], "adet":[20,35]})
hafta2 = pd.DataFrame({"gun":["Çar","Per"], "adet":[28,40]})

df_hafta = pd.concat([hafta1, hafta2], ignore_index=True)
print(df_hafta)

# 10.2: Yatay Birleştirme (Sütun Bazlı, axis=1)
gelir = pd.DataFrame({"toplam_gelir":[1000,1750]})
df_yan_yana = pd.concat([hafta1, gelir], axis=1)
print(df_yan_yana)
```

---## Bölüm 11 — GroupByAşağıdaki veriyi kullanın:

```python
df_calisan = pd.DataFrame({
    "sube": ["Kadıköy", "Beşiktaş", "Kadıköy", "Şişli", "Beşiktaş"],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya", "Gökhan"],
    "satis": [12000, 18500, 9500, 21000, 16000]
})
```

**Soru 11.1:** Şubeye göre **toplam** satışı bulun.**Soru 11.2:** Şubeye göre **ortalama** satışı bulun.**Soru 11.3:** `.agg(["sum","mean","count"])` kullanarak şube başına üç istatistiği aynı anda hesaplayın.

```python
# 11.1: Şube Bazında Toplam Satış
print(df_calisan.groupby("sube")["satis"].sum())

# 11.2: Şube Bazında Ortalama Satış
print(df_calisan.groupby("sube")["satis"].mean())

# 11.3: Çoklu Agregasyon (.agg)
print(df_calisan.groupby("sube")["satis"].agg(["sum", "mean", "count"]))
```

---## Bölüm 12 — Pivot TableAşağıdaki veriyi kullanın:

```python
df_satis = pd.DataFrame({
    "sehir": ["Bursa", "Adana", "Antalya", "Bursa", "Adana"],
    "kategori": ["Elektronik", "Giyim", "Elektronik", "Giyim", "Elektronik"],
    "satis": [5000, 3200, 4100, 2800, 3900]
})
```

**Soru 12.1:** `index="sehir"`, `columns="kategori"`, `values="satis"` ile bir pivot tablo oluşturun.**Soru 12.2:** Aynı pivot tabloyu `aggfunc="sum"` ve `fill_value=0` parametreleriyle oluşturup eksik hücrelerin nasıl doldurulduğunu gözlemleyin.

```python
# 12.1: Temel Pivot Table
pivot1 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis")
print("--- Varsayılan Pivot Table ---")
print(pivot1)

# 12.2: Aggfunc ve fill_value ile Pivot Table
pivot2 = df_satis.pivot_table(index="sehir", columns="kategori", values="satis", aggfunc="sum", fill_value=0)
print("\n--- Dolu ve Toplamlı Pivot Table ---")
print(pivot2)
```

---## Bölüm 13 — Apply ile String İşlemleri**Soru 13.1:** `df_kullanici = pd.DataFrame({"kullanici_adi": ["baran23", "CEYLIN_K", "emre.y"]})` oluşturun.**Soru 13.2:** `apply(str.lower)` kullanarak tüm kullanıcı adlarını küçük harfe çeviren bir `kullanici_adi_kucuk` sütunu ekleyin.**Soru 13.3:** `apply` ve `lambda` ile kullanıcı adının kaç karakter olduğunu gösteren bir `karakter_sayisi` sütunu ekleyin.

```python
# 13.1: DataFrame Oluşturma
df_kullanici = pd.DataFrame({"kullanici_adi": ["baran23", "CEYLIN_K", "emre.y"]})

# 13.2: Küçük Harfe Çevirme
df_kullanici['kullanici_adi_kucuk'] = df_kullanici['kullanici_adi'].apply(str.lower)

# 13.3: Karakter Sayısını Hesaplama
df_kullanici['karakter_sayisi'] = df_kullanici['kullanici_adi'].apply(lambda x: len(x))

print(df_kullanici)
```

---## Bölüm 15 — Matplotlib ile Grafik ÇizmeAşağıdaki hücreyi çalıştırarak bir yıllık çeyreklik gelir verisi oluşturun:

```python
df_gelir = pd.DataFrame({
    "ceyrek": ["Ç1", "Ç2", "Ç3", "Ç4"],
    "gelir": [42000, 55000, 48000, 61000],
    "gider": [30000, 38000, 33000, 40000]
})
```

**Soru 15.1:** `gelir` değerlerini `kind="line"` ile, üzerinde nokta işaretleri (`marker="o"`) olacak şekilde çizgi grafikle çizin; başlık ve eksen etiketleri ekleyin.**Soru 15.2:** `gelir` değerlerini çeyreklere göre `kind="bar"` ile bar grafikle çizin (renk seçiminiz size ait).**Soru 15.3:** `gelir` sütununu çeyreklerin payını gösterecek şekilde `kind="pie"` ile pasta grafikle çizin (`autopct="%1.1f%%"` kullanın).**Soru 15.4:** `gelir` ve `gider` sütunlarını `kind="scatter"` ile scatter grafikte karşılaştırın (x ekseni gelir, y ekseni gider olsun) ve grafiğin ne anlattığını bir cümleyle yorumlayın.

```python
# 15.1: Çizgi Grafiği
df_gelir.plot(x="ceyrek", y="gelir", kind="line", marker="o")
plt.title("Çeyreklik Gelir Grafiği")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.show()

# 15.2: Bar Grafiği
df_gelir.plot(x="ceyrek", y="gelir", kind="bar", color="teal")
plt.title("Çeyreklere Göre Gelirler")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir")
plt.show()

# 15.3: Pasta Grafiği
df_gelir.set_index("ceyrek")["gelir"].plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Gelirlerin Çeyreklik Dağılımı")
plt.ylabel("")  # Yan eksen yazısını temizlemek için
plt.show()

# 15.4: Scatter Grafiği ve Yorum
df_gelir.plot(x="gelir", y="gider", kind="scatter", color="red", s=50)
plt.title("Gelir vs Gider")
plt.show()

# Yorum: "Gelir arttıkça giderin de doğru orantılı olarak arttığı pozitif bir ilişki gözlemlenmektedir."
```

---## Bonus — Kapsamlı UygulamaBir online kursun satış verisini simüle edin:1. `kurs_adi`, `kategori` (örn. Yazılım, Tasarım, Pazarlama) ve `satis_adedi` sütunlarından oluşan, en az 8 satırlık bir DataFrame oluşturun.2. Kategoriye göre toplam satış adedini `groupby` ile bulun.3. En çok satan 3 kursu `.sort_values()` ile bulun.4. Kategori bazında toplam satışı gösteren bir bar grafik çizin.5. `apply` ile satış adedi 50 üzeri olanlara "Popüler", altındakilere "Standart" yazan yeni bir sütun ekleyin.

```python
# 1. En az 8 satırlık DataFrame oluşturma
df_kurs = pd.DataFrame({
    "kurs_adi": ["Python 101", "Veri Bilimi", "UI/UX Tasarım", "SEO Uzmanlığı", "Django Web", "Photoshop", "Google Ads", "React Native"],
    "kategori": ["Yazılım", "Yazılım", "Tasarım", "Pazarlama", "Yazılım", "Tasarım", "Pazarlama", "Yazılım"],
    "satis_adedi": [120, 85, 45, 30, 65, 40, 90, 25]
})

# 2. Kategoriye göre toplam satış adedi
kategori_toplam = df_kurs.groupby("kategori")["satis_adedi"].sum()
print("--- Kategori Bazlı Toplam Satış ---")
print(kategori_toplam)

# 3. En çok satan 3 kurs
top_3_kurs = df_kurs.sort_values(by="satis_adedi", ascending=False).head(3)
print("\n--- En Çok Satan 3 Kurs ---")
print(top_3_kurs)

# 4. Kategori bazında toplam satış bar grafiği
kategori_toplam.plot(kind="bar", color="orange")
plt.title("Kategori Bazında Toplam Satışlar")
plt.xlabel("Kategori")
plt.ylabel("Toplam Satış Adedi")
plt.show()

# 5. Popülerlik sütunu ekleme (apply & lambda)
df_kurs['durum'] = df_kurs['satis_adedi'].apply(lambda x: "Popüler" if x >= 50 else "Standart")
print("\n--- Güncellenmiş Kurs Tablosu ---")
print(df_kurs)
```
