```python
def sayi_durumu(s):
    if s > 0:
        return "Pozitif"
    elif s < 0:
        return "Negatif"
    else:
        return "Sıfır"

def en_buyuk_uc(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print(sayi_durumu(-7))
print("En büyük:", en_buyuk_uc(15, 42, 28))
```

```python
sepet = [
    {"urun": "Tişört", "fiyat": 80},
    {"urun": "Ayakkabı", "fiyat": 350},
    {"urun": "Pantolon", "fiyat": 150}
]

# 100 TL üzeri ürünleri listeleme
print("100 TL üzeri ürünler:")
for item in sepet:
    if item["fiyat"] > 100:
        print(f"- {item['urun']}: {item['fiyat']} TL")

# Toplam tutar ve indirim kontrolü
toplam = sum(item["fiyat"] for item in sepet)
print(f"Toplam Tutar: {toplam} TL")

if toplam > 500:
    print("%10 indirim kazandınız!")
    indirimli_fiyat = toplam * 0.90
    print(f"Ödenecek Tutar: {indirimli_fiyat} TL")
```

```python
class Araba:
    def __init__(self, marka, model, hiz=0):
        self.marka = marka
        self.model = model
        self.hiz = hiz

    def hizlandir(self, artis):
        self.hiz += artis

    def bilgi_goster(self):
        print(f"Marka: {self.marka}, Model: {self.model}, Hız: {self.hiz} km/s")

# Kullanım örneği:
araba1 = Araba("Toyota", "Corolla", 50)
araba1.hizlandir(30)
araba1.bilgi_goster()  # Çıktı: Marka: Toyota, Model: Corolla, Hız: 80 km/s
```

```python
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.uzunluk = uzunluk
        self.genislik = genislik

    def alan_hesapla(self):
        return self.uzunluk * self.genislik

    def cevre_hesapla(self):
        return 2 * (self.uzunluk + self.genislik)

# Kullanım örneği:
dikdortgen1 = Dikdortgen(10, 5)
print("Alan:", dikdortgen1.alan_hesapla())    # Çıktı: Alan: 50
print("Çevre:", dikdortgen1.cevre_hesapla())  # Çıktı: Çevre: 30
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa_sayisi):
        self.ad = ad
        self.yazar = yazar
        self.sayfa_sayisi = sayfa_sayisi

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"

# Kullanım örneği:
kitap1 = Kitap("Suç ve Ceza", "Dostoyevski", 671)
print(kitap1.ozet())  # Çıktı: Suç ve Ceza - Dostoyevski (671 sayfa)
```
