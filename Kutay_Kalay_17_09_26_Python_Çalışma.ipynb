```python
data = {
    "isim": ["Ali", "Veli", "Ayşe", "Fatma", "Mehmet", "Zeynep"],
    "yas": [19, 26, 32, 18, 45, 29],
    "maas": [18000, 25000, 31000, 15000, 48000, 28000]
}
df = pd.DataFrame(data)

print("İlk 3 Satır:\n", df.head(3), "\n")
print("Son 2 Satır:\n", df.tail(2), "\n")
print("DataFrame Boyutu:", df.shape, "\n")

print("Veri Tipleri ve Bilgi:")
df.info()
print("\nİstatistiksel Özet:\n", df.describe())
```

```python
df = pd.DataFrame({
    "isim": ["Ali", "Veli", "Ayşe", "Fatma", "Mehmet"],
    "yas": [19, 26, 32, 18, 45],
    "maas": [18000, 25000, 31000, 15000, 48000]
})

# 1. Tek sütun seçimi
print("Sadece İsim Sütun:\n", df["isim"], "\n")

# 2. Birden fazla sütun seçimi
print("İsim ve Yaş:\n", df[["isim", "yas"]], "\n")

# 3. .loc[] ile ilk 3 satır (etiket tabanlı, 0, 1 ve 2 dahil)
print("İlk 3 Satır (.loc):\n", df.loc[0:2], "\n")

# 4. .iloc[] ile 2. satır, 2. sütun (indeksler 0'dan başladığı için [1, 1])
print("2. Satır, 2. Sütun Kesişimi:", df.iloc[1, 1])
```

```python
df = pd.DataFrame({
    "isim": ["Ali", "Veli", "Ayşe", "Fatma", "Mehmet"],
    "yas": [19, 26, 32, 17, 42],
    "sehir": ["İstanbul", "Ankara", "İstanbul", "İzmir", "Ankara"],
    "maas": [18000, 25000, 31000, 15000, 48000]
})

# 1. Yaşı 25'ten büyük olanlar
print("Yaş > 25:\n", df[df["yas"] > 25], "\n")

# 2. Maaşı 20000 - 30000 arasında olanlar (&)
print("Maaş 20k-30k Arası:\n", df[(df["maas"] >= 20000) & (df["maas"] <= 30000)], "\n")

# 3. Belirli bir şehirde yaşayanlar
print("İstanbul'da Yaşayanlar:\n", df[df["sehir"] == "İstanbul"], "\n")

# 4. Yaşı < 20 veya > 40 olanlar (|)
print("Yaş < 20 veya > 40:\n", df[(df["yas"] < 20) | (df["yas"] > 40)])
```

```python
df = pd.DataFrame({
    "isim": ["Ali", "Veli", "Ayşe", "Fatma"],
    "yas": [17, 26, 32, 15],
    "maas": [18000, 25000, 31000, 15000]
})

# 1. Hesaplanan sütun ekleme
df["dogum_yili"] = 2026 - df["yas"]

# 2. Sütun değerini %10 artırma
df["maas"] = df["maas"] * 1.10

# 3. apply() ve lambda ile koşullu sütun oluşturma
df["durum"] = df["yas"].apply(lambda y: "Yetişkin" if y >= 18 else "Çocuk")

print(df)
```

```python
df = pd.DataFrame({
    "isim": ["Ali", "Veli", "Ayşe", "Fatma", "Mehmet"],
    "sehir": ["Ankara", "İstanbul", "Ankara", "İzmir", "İstanbul"],
    "yas": [25, 22, 30, 19, 22],
    "maas": [20000, 35000, 28000, 15000, 40000]
})

# 1. Yaşa göre küçükten büyüğe
print("Yaşa Göre (Artan):\n", df.sort_values(by="yas"), "\n")

# 2. Maaşa göre büyükten küçüğe
print("Maaşa Göre (Azalan):\n", df.sort_values(by="maas", ascending=False), "\n")

# 3. Önce şehre (A-Z), sonra yaşa göre (küçükten büyüğe) sıralama
print("Şehir ve Yaşa Göre:\n", df.sort_values(by=["sehir", "yas"]))
```

```python
#MİNİ ÖDEV BONUS

# 1. DataFrame oluşturma
sinif_df = pd.DataFrame({
    "isim": ["Ahmet", "Berna", "Cem", "Deniz", "Elif", "Fırat"],
    "yas": [21, 23, 20, 22, 24, 21],
    "sehir": ["İstanbul", "Ankara", "İzmir", "İstanbul", "Ankara", "Bursa"],
    "ders1": [85, 60, 90, 55, 70, 95],
    "ders2": [75, 65, 80, 60, 85, 90],
    "ders3": [80, 70, 85, 50, 75, 85]
})

# 2. Ders ortalamasını hesaplama
sinif_df["ortalama"] = sinif_df[["ders1", "ders2", "ders3"]].mean(axis=1)

# 3. Ortalaması 70'in üzerinde olanları filtreleme
basarili_ogrenciler = sinif_df[sinif_df["ortalama"] > 70]

# 4. Yaşa göre küçükten büyüğe sıralama
sonuc = basarili_ogrenciler.sort_values(by="yas")

print(sonuc[["isim", "yas", "sehir", "ortalama"]])
```
